```python
#set1 q1
#a stdunt detsil arre stored in a dictionarycontaining name roll no. and marks in three subs maths stats python write a python program to calculate the total and precentage determine the result (pass if every subject mark is atleast 40 overwise fail)and displayneatly formatted marksheet using f strings with aligned columns



student = {
    "name": "Alice Johnson",
    "roll_no": "STU2023041",
    "marks": {
        "maths": 85,
        "stats": 67,
        "python": 78
    }
}


student_name = student["name"]
roll = student["roll_no"]
maths_mark = student["marks"]["maths"]
stats_mark = student["marks"]["stats"]
python_mark = student["marks"]["python"]


total = maths_mark + stats_mark + python_mark
percentage = (total / 3)*1


if maths_mark >= 40 and stats_mark >= 40 and python_mark >= 40:
    result = "PASS"
else:
    result = "FAIL"



print("              STUDENT MARKSHEET              ")

print(f"Name    : {student_name}")
print(f"Roll No : {roll}")
print(f"Subject             Max Marks   Marks Obtained")
print("---------------------------------------------")
print(f"Maths               100         {maths_mark}")
print(f"Stats               100         {stats_mark}")
print(f"Python              100         {python_mark}")
print(f"Total               300         {total}")
print(f"Percentage                      {percentage}%")
print(f"Result                          {result}")
```

```python
#set1 q2
#write a function income_tax income thaht returns the tax payable using these practice slabs upto 3L-nil next 4L(upto 7L)-5% the next 3l (upto 10l)-10% above 10l 20% tax must be calculated slab wise accept an income print the tax and effective tax rate(tax/icome as a percentage , 2 decimals)


income = float(input("Enter your income: "))


if income <= 300000:
    tax = 0.0
elif income <= 700000:
    tax = (income - 300000) * 0.05
elif income <= 1000000:
    tax = 20000 + (income - 700000) * 0.10
else:
    tax = 20000 + 30000 + (income - 1000000) * 0.20


rate = (tax / income) * 100

print("Income:", income)
print("Tax Payable:", tax)
print("Effective Tax Rate:", rate, "%")
```
