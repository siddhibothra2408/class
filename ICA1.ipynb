```python
#-------------question 1 set 4--------------
sentence=input("enter senstence:")
vowel=0
constant=0
digits=0
space=0
for ch in sentence:
    if (ch =='a' or ch=='e' or ch=='o' or ch=='i' or ch=='u'):
        vowel = vowel + 1
    elif (ch=='A' or ch=='E' or ch=='I' or ch=='O' or ch=='U'):
        vowel = vowel + 1
    elif (ch >='0' and ch <='9'):
        digits = digits + 1
    elif ch==' ':
        space = space + 1
    else:
        constant = constant + 1

print("vowel=", vowel)
print("constant=", constant)
print("digits=", digits)
print("blank spaces=", space)
```
